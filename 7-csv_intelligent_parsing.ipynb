```python
from langchain_core.documents import Document

def parse_custom_csv(file_path):
    # Read CSV file with pandas
    df = pd.read_csv(file_path)

    documents = []
    for index, row in df.iterrows():
        # Create a document for each row

        row_dict = row.to_dict()  # Convert the row to a dictionary

        content = str(row_dict)  # Convert the dictionary to a string representation

        metadata = {"row_index": index, "file_path": file_path, "file_type": "csv"}  # Add row index and file path as metadata

        document = Document(page_content=content, metadata=metadata)
        documents.append(document)

    return documents
```

```python
csv_data = parse_custom_csv("data/csv/employees.csv")
print(f"csv_data: {csv_data}")
```
