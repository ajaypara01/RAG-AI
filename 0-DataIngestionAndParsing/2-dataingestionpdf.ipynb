# PDF Loading and Splitting

This notebook loads text from a PDF and explores text splitters. PDF extraction makes document content available to a RAG pipeline, while chunking creates passages small enough to embed and retrieve. Check extracted text quality because PDF layout can affect reading order and formatting.

### PDF file Loader - PyMuPDFLoader

```python
from langchain_community.document_loaders import (
    PyMuPDFLoader,
    PyPDFLoader,
)

pdf_loader = PyMuPDFLoader("../data/pdf/Ajay-Sharma.pdf")
pdf_documents = pdf_loader.load()
print(f"PDF Documents loaded: {len(pdf_documents)}")
print(f"Document: {pdf_documents}")
print(f"\nDocument Content: {pdf_documents[0].page_content[:500]} ......")
```

```python
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter
    )
try:
    pdf_text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=200,
        chunk_overlap=20,
        separators=["\n\n", "\n", " ", ""],
    )
    pdf_text_splitter = CharacterTextSplitter(
        chunk_size=200,
        chunk_overlap=20,
        separator="\n",
    )

    pdf_chunks = pdf_text_splitter.split_text(pdf_documents[0].page_content)
    print(f"Number of chunks created: {len(pdf_chunks)}")

    print(f"First chunk: {pdf_chunks[0]} ......")
except Exception as e:
        print(f"Error creating RecursiveCharacterTextSplitter: {e}")
```
