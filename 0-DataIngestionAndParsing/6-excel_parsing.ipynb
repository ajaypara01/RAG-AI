# Excel Parsing

This notebook creates and parses Excel workbook content. Converting spreadsheet data into documents makes it usable in document-oriented LangChain workflows; inspect the loader output to verify how sheets, rows, and cell values are represented.

### Excel parsing and Creation

```python
import pandas as pd
```

### Create directory and Excel file

```python
employees = [
    {
        "id": 1,
        "name": "Ajay",
        "role": "Frontend Developer",
        "salary": 80000
    },
    {
        "id": 2,
        "name": "Rahul",
        "role": "Backend Developer",
        "salary": 90000
    },
    {
        "id": 3,
        "name": "Priya",
        "role": "QA Engineer",
        "salary": 70000
    }
]

df = pd.DataFrame(employees)
print(df)
```

```python
import os
os.makedirs("../data/excel", exist_ok=True)
df.to_excel("../data/excel/employees.xlsx", index=False)
```

### EXCEl Data Parsing

```python
from langchain_community.document_loaders import UnstructuredExcelLoader

excel_loader = UnstructuredExcelLoader("../data/excel/employees.xlsx", mode="elements")

documents = excel_loader.load()

print(f"Excel Documents loaded: {len(documents)}")
for i, doc in enumerate(documents[:4]):
    print(f"\nDocument {i+1} Content: {doc.page_content[:100]} ......")
    print(f"Document metadata: {doc.metadata}")
```
