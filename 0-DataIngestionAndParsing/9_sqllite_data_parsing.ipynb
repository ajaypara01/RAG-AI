This notebook connects to a SQLite database and explores loading or querying its data through LangChain utilities. It demonstrates bringing relational data into an LLM workflow; use structured SQL queries when exact filtering and joins matter, and convert results to documents when document-style retrieval is appropriate.

### SQLite parser


```python
import sqlite3
import os

# Create Database folder if it doesn't exist

os.makedirs("../data/database", exist_ok=True)
```

```python
# Create Schema company.db and table employees if they don't exist

conn = sqlite3.connect("../data/database/company.db")
cursor = conn.cursor()
```

```python
# Create Table Employee

cursor.execute('''CREATE TABLE IF NOT EXISTS employees (id INTEGER PRIMARY KEY, name TEXT, role TEXT, department TEXT, salary REAL)''')

cursor.execute(''' CREATE TABLE IF NOT EXISTS department (id INTEGER PRIMARY KEY, project_name TEXT, desc TEXT)''')
```

```python
# Insert sample rows into both tables. Run this cell after creating the tables.
employees = [
    ("John Doe", "Software Engineer", "Engineering", 85000.00),
    ("Jane Smith", "Product Manager", "Product", 92000.00),
    ("Alex Johnson", "Data Scientist", "Research", 98000.00),
]

projects = [
    ("AI Research", "Developing machine learning models and AI algorithms."),
    ("Mobile App Development", "Building a cross-platform mobile application."),
    ("Data Platform", "Creating a centralized platform for analytics data."),
]

cursor.executemany(
    "INSERT INTO employees (name, role, department, salary) VALUES (?, ?, ?, ?)",
    employees,
)
cursor.executemany(
    'INSERT INTO department (project_name, "desc") VALUES (?, ?)',
    projects,
)


cursor.execute("SELECT * FROM employees")
conn.commit()
conn.close()
```
