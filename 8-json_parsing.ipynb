### JSON PARSING - 

```python
import os
import json

os.makedirs("data/json", exist_ok=True)

company_data = {
    "company_name": "Tech Innovators Inc.",
    "company_address": "123 Innovation Drive, Tech City, TC 45678",
    "employees": [
        {
            "name": "John Doe",
            "position": "Software Engineer",
            "email": "john.doe@techinnovators.com",
        },
        {
            "name": "Jane Smith",
            "position": "Product Manager",
            "email": "jane.smith@techinnovators.com",
        },
    ],
    "projects": [
        {
            "project_name": "AI Research",
            "description": "Developing cutting-edge AI algorithms.",
            "start_date": "2023-01-15",
            "end_date": "2024-06-30",
        },
        {
            "project_name": "Mobile App Development",
            "description": "Creating a cross-platform mobile application.",
            "start_date": "2023-03-01",
            "end_date": "2024-12-31",
        },
    ],
}

with open("data/json/company_data.json", "w") as json_file:
    json.dump(company_data, json_file, indent=4)
```

```python
import json
from typing import List
from langchain_core.documents import Document


def parse_custom_json(file_path: str) -> List[Document]:
    """Parse company JSON into one Document per employee and project."""
    with open(file_path, "r") as json_file:
        data = json.load(json_file)

    company_metadata = {
        "company_name": data.get("company_name"),
        "company_address": data.get("company_address"),
        "file_type": "json",
    }
    documents = []

    for employee in data.get("employees", []):
        employee_metadata = {
            **company_metadata,
            "record_type": "employee",
            **employee,
        }
        employee_content = "Employee\n" + "\n".join(
            f"{key.replace('_', ' ').title()}: {value}"
            for key, value in employee.items()
        )
        documents.append(
            Document(page_content=employee_content, metadata=employee_metadata)
        )

    for project in data.get("projects", []):
        project_metadata = {
            **company_metadata,
            "record_type": "project",
            **project,
        }
        project_content = "Project\n" + "\n".join(
            f"{key.replace('_', ' ').title()}: {value}"
            for key, value in project.items()
        )
        documents.append(
            Document(page_content=project_content, metadata=project_metadata)
        )

    return documents


json_content = parse_custom_json("data/json/company_data.json")

for document in json_content:
    print(f"Metadata: {document.metadata}")
    print(f"Content:\n{document.page_content}\n")

print(f"JSON document count: {len(json_content)}")
```
