# Word Document Parsing

This notebook explores extracting content from Word documents with different LangChain loaders. Choose a loader based on whether plain text is enough or document elements and structure matter; the resulting content can then be split, embedded, and retrieved.

### Docx file parsing

```python
from langchain_community.document_loaders import (Docx2txtLoader, UnstructuredWordDocumentLoader)
```

```python
docs_loader = Docx2txtLoader("../data/docs/js_interview.docx")
docsuments = docs_loader.load()
print(f"Documents loaded: {len(docsuments)}")
print(f"Document Content: {" ".join(docsuments[0].page_content[:100].split())} ......")
```

```python
unstructurred_loader = UnstructuredWordDocumentLoader("../data/docs/js_interview.docx", mode="elements")
unstructured_documents = unstructurred_loader.load()
print(f"Unstructured Documents loaded: {len(unstructured_documents)}")
for i, doc in enumerate(unstructured_documents[:4]):
    print(f"\nDocument {i+1} Content: {doc.page_content[:100]} ......")
    print(f"Document metadata: {doc.metadata}")
```
