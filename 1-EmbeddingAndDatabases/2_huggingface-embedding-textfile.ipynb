# Embedding Text from a File

This notebook loads a text file, splits its content, and creates embeddings for the resulting text pieces. Splitting keeps each vector focused on a manageable passage, while the embeddings make semantic similarity search possible. This demonstrates the bridge between file ingestion and vector indexing.

### Read text files from data/text_files

```python
from langchain_community.document_loaders import TextLoader

text_load = TextLoader("../data/texts_files/langchain_info.txt")
documents = text_load.load()
print(f"Documents length: {len(documents)}")
print(f"Document: {documents[0].page_content}")
```

```python
# split the content

from langchain_text_splitters import CharacterTextSplitter

split_data = CharacterTextSplitter(
    separator=".",
    chunk_size=100,
    chunk_overlap=25
)

print(documents[0].page_content)
chunk_data = split_data.split_text(documents[0].page_content)

print(f"Chunk length {len(chunk_data)}")
print(f"Chunk data: {chunk_data}")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L12-v2"
)

embedded_documents = embedding.embed_documents(chunk_data)

print(f"Documents size: {len(embedded_documents[1])}")
print(f"Embedded for docs: {embedded_documents[0]}")
print(f"Text docs 0 : {chunk_data[0]}")
```
