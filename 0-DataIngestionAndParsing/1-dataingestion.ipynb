### TextLoader - Single file loader

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/texts_files/rag_info.txt", encoding="utf-8")

documents = loader.load()

print(f"Documents loaded: {len(documents)}")
print(type(documents))
print(f"Document Content: {documents[0].page_content[:100]} ......")
print(f"Document MetaData: {documents[0].metadata}")
```

### DirectoryLoader - multiple file loader

```python
from langchain_community.document_loaders import DirectoryLoader

loader = DirectoryLoader(
 "../data/texts_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=True
)

documents = loader.load()

print(f"Documents loaded: {len(documents)}")

for i, doc in enumerate(documents):
    print(f"\nDocuments {i + 1}:")
    print(f"Content {doc.page_content[:100]} ......")
    print(f"Document MetaData: {doc.metadata}")
```

### CharactorTextSplitters

```python
print(documents[0].page_content)
text = documents[0].page_content
char_splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=15,
    length_function=len,
    separator="."
)

char_chunks = char_splitter.split_text(text)

print(f"Chunks Length: {len(char_chunks)}")
print(f"chunk 1: {char_chunks[0]}")

print(f"Chunks: {char_chunks}")

```

### Recursive Charactor Splitter

```python
recursive_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " ", ""],
    chunk_size=100,
    chunk_overlap=15,
    length_function=len
)

recursive_chunks = recursive_splitter.split_text(text)

print(f"Recursive Chunks Length: {len(recursive_chunks)}")
print(f"Recursive chunk 1: {recursive_chunks[0]}")
print(f"Recursive Chunks: {recursive_chunks}")
```
