# Sentence Embeddings

This notebook converts sentences into numeric vectors with a Hugging Face sentence-transformer model. Embeddings represent semantic features and can be compared for similarity; they are not generated explanations or human-readable text. This is the foundation for semantic search and vector-store retrieval.

### Basic Sentence convert into embedding vectors

```python
from langchain_huggingface import HuggingFaceEmbeddings

sentence = "Hi how are you, are you feeling well ?"

embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L12-v2",

)
embedded = embedding.embed_query(sentence)
```

```python
print(f"Sentense: {sentence}")
print(f"Embedding length: {len(embedded)}")
print(f"Embeddings: {embedded}")
```
