### Smart PDF Ingestion - Will Use Smart PDF class to put validations and text cleanup

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from typing import List
from langchain_core.documents import Document
```

### Smart PDF Class

```python
class SmartPDFIngestor:
    def __init__(self, chunk_size: int = 1000, chunk_overlap: int = 100):
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size = self.chunk_size,
            chunk_overlap = self.chunk_overlap,
            separators = ["\n\n", "\n", " ", ""]
        )

    def process_pdf(self, file_path: str) -> List[Document]:
        loader = PyPDFLoader(file_path)
        pages = loader.load()

        chunks_list = []
        for page_num, page in enumerate(pages):
            # Clean the page content by removing extra whitespace
            cleaned_content = self._clean_text(page.page_content)

            chunks = self.text_splitter.create_documents(
                texts = [cleaned_content],
                metadatas = [
                    {
                        **page.metadata,
                        "source": file_path,
                        "page": page_num + 1,
                        "total_pages": len(pages),
                        "splitter": "RecursiveCharacterTextSplitter",
                    }
                ]
            )
            chunks_list.extend(chunks)
        return chunks_list
    
    def _clean_text(self, text: str) -> str:
        # Remove extra whitespace and normalize the text
        return " ".join(text.split())
```

```python
processor = SmartPDFIngestor(chunk_size=500, chunk_overlap=30)
```

```python
processed = processor.process_pdf("data/pdf/Ajay-Sharma.pdf")
print(f"Processed Chunks: {len(processed)}")

print(f"Chunk Content: {processed[0].page_content[:500]} ......")
```
