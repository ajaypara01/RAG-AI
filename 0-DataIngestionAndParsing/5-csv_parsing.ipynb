# CSV Parsing

This notebook creates and loads CSV data as LangChain documents. A CSV loader typically turns rows into document content with metadata, making table records available to downstream search and RAG. Preserve row identifiers when answers need to be traced back to the source.

### CSV Creatioin And Data Parsing

```python
import pandas as pd
```

```python
employees = [
    {
        "id": 1,
        "name": "Ajay",
        "role": "Frontend Developer",
        "salary": 80000
    },
    {
        "id": 2,
        "name": "Rahul",
        "role": "Backend Developer",
        "salary": 90000
    },
    {
        "id": 3,
        "name": "Priya",
        "role": "QA Engineer",
        "salary": 70000
    }
]

df = pd.DataFrame(employees)
print(df)
```

### CSV File Creation

```python
import os

os.makedirs("../data/csv", exist_ok=True)
df.to_csv("../data/csv/employees.csv", index=False)
```

### CSV Data Parsing

```python
from langchain_community.document_loaders import CSVLoader

csv_loader = CSVLoader(file_path="../data/csv/employees.csv", encoding="utf-8")

documents = csv_loader.load()

print(f"CSV Documents loaded: {len(documents)}")
print(f"CSV Document Content: {documents} ......")
```
